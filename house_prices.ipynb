# Data Preprocessing

```python
import numpy as np
from sklearn.model_selection import train_test_split

# Read the data
X_full = pd.read_csv('./Data/train.csv', index_col='Id')
X_test_full = pd.read_csv('./Data/test.csv', index_col='Id')

# Remove rows with missing target, separate target from predictors
X_full.dropna(axis=0, subset=['SalePrice'], inplace=True)
y = X_full.SalePrice
X_full.drop(['SalePrice'], axis=1, inplace=True)

# Break off validation set from training data
X_train_full, X_valid_full, y_train, y_valid = train_test_split(X_full, y, 
                                                                train_size=0.8, test_size=0.2,
                                                                random_state=0)

# "Cardinality" means the number of unique values in a column
# Check high cardinality columns (categorical columns represented by str)
high_cardinality_columns = [cname for cname in X_train_full.select_dtypes(include=["string"]).columns if
                        X_train_full[cname].nunique() > 10]

# for column in high_cardinality_columns:
#     print(f"{column}:{X_train_full[column].unique()}")

# All high cardinality columns contains potentially useful information, do not dorp
```

```python
from sklearn.base import BaseEstimator, TransformerMixin

# Define transformer to drop columns with high percentage of Nan values.
class DropColumnsWithHighMissingValuesTransformer(BaseEstimator,TransformerMixin):
    def __init__(self,threshold):
        self.threshold = threshold
        self.columns_to_drop = None

    def fit(self,X: pd.DataFrame, y = None):
        missing_values = X.isna().mean()

        self.columns_to_drop = missing_values[missing_values>self.threshold].index
        print(f"Columns with high missing values: {list(self.columns_to_drop)}")
        return self

    def transform(self, X: pd.DataFrame):
        print(f"Pre-transform X shape: {X.shape}")

        if self.columns_to_drop is not None:
            X = X.drop(columns=self.columns_to_drop, errors= 'ignore')
            print(f"Post-transform X shape: {X.shape}, removed columns {list(self.columns_to_drop)}")
     
        return X

drop_columns_transformer = DropColumnsWithHighMissingValuesTransformer(0.2)

# Drop columns with high percentage of Nan values.
X_train_drop = drop_columns_transformer.fit_transform(X_train_full)
X_valid_drop = drop_columns_transformer.transform(X_valid_full)
X_test_drop = drop_columns_transformer.transform(X_test_full)

# Define categorical and numerical columns
categorical_cols = [cname for cname in X_train_drop.columns if
                    X_train_full[cname].dtype in ["string"]]

numerical_cols = [cname for cname in X_train_drop.columns if 
                X_train_full[cname].dtype in ['int64', 'float64']]

my_cols = categorical_cols + numerical_cols

X_train = X_train_drop[my_cols].copy()
X_valid = X_valid_drop[my_cols].copy()
X_test = X_test_drop[my_cols].copy()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler


# Preprocessing for numerical data
numerical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
],verbose=True)

# Preprocessing for categorical data
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore',sparse_output=False))
],verbose=True)

# Bundle preprocessing for numerical and categorical data
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_cols),
        ('cat', categorical_transformer, categorical_cols)
    ])

# Bundle preprocessing in a pipeline
final_pipeline = Pipeline(steps=[('preprocessor', preprocessor)])

# Preprocessing of training data, fit model 
final_pipeline.fit(X_train)
print(final_pipeline.get_feature_names_out)



```

```python
# Transform the features in training and validation set
X_train_transformed =  final_pipeline.transform(X_train)
X_valid_transformed =  final_pipeline.transform(X_valid)
X_test_transformed =  final_pipeline.transform(X_test)

#print(pd.DataFrame(X_train_transformed).describe())
#print(pd.DataFrame(X_train_transformed))
#print(pd.DataFrame(X_valid_transformed).describe())
#print(pd.DataFrame(X_valid_transformed))
print(pd.DataFrame(X_test_transformed))
```

# Model Evaluation

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import root_mean_squared_error,root_mean_squared_log_error

# Preprocessing of validation data, get predictions
y_prediction_training = np.maximum(0,model.predict(X_train_transformed))
y_prediction_validation = np.maximum(0,model.predict(X_valid_transformed))

#print('MAE:', mean_absolute_error(y_valid, preds))
print('RMSLE', root_mean_squared_log_error(y_train,y_prediction_training))
print('RMSLE', root_mean_squared_log_error(y_valid,y_prediction_validation))
```

```python
from sklearn.linear_model import Ridge

# Validation RMSLE is way larger than training RMSLE, suggesting overfitting.
model_ridge = Ridge()

model_ridge.alpha = 1
model_ridge.fit(X_train_transformed,y_train)
print(f'Ridge regression model has trained successfully')

y_prediction_training_ridge = np.maximum(0,model_ridge.predict(X_train_transformed))
y_prediction_validation = np.maximum(0,model_ridge.predict(X_valid_transformed))


print('RMSLE', root_mean_squared_log_error(y_train,y_prediction_training_ridge))
print('RMSLE', root_mean_squared_log_error(y_valid,y_prediction_validation))

# increase alpha
model_ridge.alpha = 15
model_ridge.fit(X_train_transformed,y_train)
print(f'Ridge regression model has trained successfully')

y_prediction_training_ridge = np.maximum(0,model_ridge.predict(X_train_transformed))
y_prediction_validation = np.maximum(0,model_ridge.predict(X_valid_transformed))


print('RMSLE', root_mean_squared_log_error(y_train,y_prediction_training_ridge))
print('RMSLE', root_mean_squared_log_error(y_valid,y_prediction_validation))

# Validation RMSLE improves significantly with increased alpha.
```

```python
# iterate through different alpha to find best alpha

alpha_values = np.logspace(-3, 3)

alpha_training_tuple = []
alpha_validation_tuple = []

for alpha_value in alpha_values:
    model_ridge.alpha = alpha_value
    model_ridge.fit(X_train_transformed,y_train)

    y_prediction_training_ridge = np.maximum(0,model_ridge.predict(X_train_transformed))
    y_prediction_validation = np.maximum(0,model_ridge.predict(X_valid_transformed))

    print('alpha value',alpha_value)
    print('RMSLE Training -----------', root_mean_squared_log_error(y_train,y_prediction_training_ridge))
    print('RMSLE Validation -----------', root_mean_squared_log_error(y_valid,y_prediction_validation))

    alpha_training_tuple.append((alpha_value,root_mean_squared_log_error(y_train,y_prediction_training_ridge)))
    alpha_validation_tuple.append((alpha_value,root_mean_squared_log_error(y_valid,y_prediction_validation)))

min_training_alpha = min(alpha_training_tuple,key = lambda t:t[1])
min_validation_alpha = min(alpha_validation_tuple,key = lambda t:t[1])
print(min_training_alpha)
print(min_validation_alpha)
```
